# DROP Project Usage Notebook

This notebook demonstrates how to use the project with the **mock backend** so it runs without a real API key.

```python
import sys, os
from pathlib import Path
project_root = Path.cwd().resolve()
if project_root.name != "drop_concept_portfolio_project":
    project_root = Path("/mnt/data/drop_concept_portfolio_project")
sys.path.insert(0, str(project_root / "src"))
print(project_root)
```

```python
from drop_portfolio.config import Settings
from drop_portfolio.backends.mock import MockBackend
from drop_portfolio.services.session_service import SessionService

settings = Settings(openrouter_api_key="demo-key", runs_dir=project_root / "demo_runs")
service = SessionService(settings, MockBackend())
print("Service ready")
```

```python
session = service.create_session(
    project_brief="A gathering about things that become meaningful when we return to them.",
    initial_context="small intimate cultural gathering",
    desired_feeling="reflective, warm, quietly curious",
    seed="returning changes us",
    previous_ideas=[]
)
session
```

```python
session = service.generate_concepts(session.session_id)
[c.model_dump() for c in session.concepts]
```

```python
session = service.respond_to_concepts(
    session.session_id,
    action="refine",
    liked_concept_ids=["concept_02"],
    feedback="I like concept 2, but make it less nostalgic and more social."
)
[c.model_dump() for c in session.concepts]
```

```python
session = service.approve_concept(session.session_id, "concept_02")
session.approved_concept.model_dump()
```

```python
session = service.build_portfolio(session.session_id)
session.portfolio.model_dump()
```

```python
from pathlib import Path
run_dir = Path(session.run_dir)
print((run_dir / "final_report.md").read_text(encoding="utf-8"))
```
